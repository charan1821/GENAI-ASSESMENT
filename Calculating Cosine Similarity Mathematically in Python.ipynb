```python
{
    "cells": [
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "# Calculating Cosine Similarity Mathematically in Python\n",
    "\n",
    "\n",
    "\n",
    "### What you will learn\n",
    "1. The math formula behind Cosine Similarity\n",
    "2. Dot Product — calculated manually\n",
    "3. Magnitude (Vector Length) — calculated manually\n",
    "4. Cosine Similarity — built step-by-step from scratch\n",
    "5. Verifying our manual result using NumPy\n",
    "6. Verifying our manual result using scikit-learn\n",
    "7. Testing on real sentence embeddings (optional, intermediate)"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 1. The Math Formula\n",
    "\n",
    "Cosine similarity measures the angle between two vectors, **not** their length.\n",
    "\n",
    "$$\n",
    "\\text{cosine\\_similarity}(A, B) = \\frac{A \\cdot B}{\\|A\\| \\times \\|B\\|}\n",
    "$$\n",
    "\n",
    "Where:\n",
    "- $A \\cdot B$ = **Dot product** of vectors A and B\n",
    "- $\\|A\\|$ = **Magnitude (length)** of vector A = $\\sqrt{a_1^2 + a_2^2 + ... + a_n^2}$\n",
    "- $\\|B\\|$ = Magnitude of vector B\n",
    "\n",
    "**Result range:**\n",
    "- `1` → vectors point in the exact same direction (very similar)\n",
    "- `0` → vectors are perpendicular (unrelated)\n",
    "- `-1` → vectors point in opposite directions (opposite meaning)"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### 2. Define Two Simple Vectors\n",
    "We start with small, easy-to-check numbers before moving to real data."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "metadata": {},
   "outputs": [],
   "source": [
    "vector_A = [2, 4, 1]\n",
    "vector_B = [1, 3, 5]\n",
    "\n",
    "print(\"Vector A:\", vector_A)\n",
    "print(\"Vector B:\", vector_B)"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### 3. Step 1 — Dot Product (from scratch)\n",
    "\n",
    "Formula: $A \\cdot B = a_1 b_1 + a_2 b_2 + ... + a_n b_n$"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "metadata": {},
   "outputs": [],
   "source": [
    "def dot_product(A, B):\n",
    "    total = 0\n",
    "    for a, b in zip(A, B):\n",
    "        total += a * b\n",
    "    return total\n",
    "\n",
    "dp = dot_product(vector_A, vector_B)\n",
    "print(\"Dot Product (A . B):\", dp)"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### 4. Step 2 — Magnitude of a Vector (from scratch)\n",
    "\n",
    "Formula: $\\|A\\| = \\sqrt{a_1^2 + a_2^2 + ... + a_n^2}$"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "metadata": {},
   "outputs": [],
   "source": [
    "import math\n",
    "\n",
    "def magnitude(A):\n",
    "    total = 0\n",
    "    for a in A:\n",
    "        total += a ** 2\n",
    "    return math.sqrt(total)\n",
    "\n",
    "mag_A = magnitude(vector_A)\n",
    "mag_B = magnitude(vector_B)\n",
    "\n",
    "print(\"Magnitude of A:\", round(mag_A, 4))\n",
    "print(\"Magnitude of B:\", round(mag_B, 4))"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### 5. Step 3 — Cosine Similarity (from scratch)\n",
    "Now combine both formulas."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "metadata": {},
   "outputs": [],
   "source": [
    "def cosine_similarity_manual(A, B):\n",
    "    dp = dot_product(A, B)\n",
    "    mag_A = magnitude(A)\n",
    "    mag_B = magnitude(B)\n",
    "    similarity = dp / (mag_A * mag_B)\n",
    "    return similarity\n",
    "\n",
    "result_manual = cosine_similarity_manual(vector_A, vector_B)\n",
    "print(\"Cosine Similarity (Manual Calculation):\", round(result_manual, 4))"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "**Discussion Point:** We just calculated cosine similarity using **only** basic Python — no libraries. This is exactly what happens internally inside NumPy or scikit-learn functions."
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### 6. Verify Using NumPy\n",
    "NumPy gives the same answer, but with cleaner, faster code."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "metadata": {},
   "outputs": [],
   "source": [
    "import numpy as np\n",
    "\n",
    "A = np.array(vector_A)\n",
    "B = np.array(vector_B)\n",
    "\n",
    "dot = np.dot(A, B)\n",
    "norm_A = np.linalg.norm(A)\n",
    "norm_B = np.linalg.norm(B)\n",
    "\n",
    "result_numpy = dot / (norm_A * norm_B)\n",
    "print(\"Cosine Similarity (NumPy):\", round(result_numpy, 4))"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### 7. Verify Using scikit-learn\n",
    "This is the function used in real-world ML pipelines."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "metadata": {},
   "outputs": [],
   "source": [
    "!pip install -q scikit-learn"
   ]
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "metadata": {},
   "outputs": [],
   "source": [
    "from sklearn.metrics.pairwise import cosine_similarity\n",
    "\n",
    "# sklearn expects 2D arrays: shape (1, n)\n",
    "A_2d = np.array(vector_A).reshape(1, -1)\n",
    "B_2d = np.array(vector_B).reshape(1, -1)\n",
    "\n",
    "result_sklearn = cosine_similarity(A_2d, B_2d)[0][0]\n",
    "print(\"Cosine Similarity (scikit-learn):\", round(result_sklearn, 4))"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### 8. Compare All Three Results\n",
    "\n",
    "All three methods should give the **exact same number**, proving our manual math was correct."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "metadata": {},
   "outputs": [],
   "source": [
    "print(\"Manual     :\", round(result_manual, 6))\n",
    "print(\"NumPy      :\", round(result_numpy, 6))\n",
    "print(\"scikit-learn:\", round(result_sklearn, 6))\n",
    "\n",
    "assert round(result_manual, 4) == round(result_numpy, 4) == round(result_sklearn, 4)\n",
    "print(\"\\nAll three methods match!\")"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### 9. Try More Examples (Intermediate)\n",
    "\n",
    "Let's test edge cases to build intuition."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "id": "a05f1bb6",
   "metadata": {},
   "outputs": [],
   "source": []
  },
  {
   "cell_type": "code",
   "execution_count": 4,
   "metadata": {},
   "outputs": [],
   "source": [
    "examples = {\n",
    "    \"Identical vectors\": ([1, 2, 3], [1, 2, 3]),\n",
    "    \"Opposite vectors\": ([1, 2, 3], [-1, -2, -3]),\n",
    "    \"Perpendicular vectors\": ([1, 0], [0, 1]),\n",
    "    \"Scaled vectors (same direction)\": ([1, 2, 3], [2, 4, 6]),\n",
    "}\n",
    "\n",
    "for name, (a, b) in examples.items():\n",
    "    sim = cosine_similarity_manual(a, b)\n",
    "    print(f\"{name:35s} -> {round(sim, 4)}\")"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "**Discussion Point:**\n",
    "- Identical vectors → similarity = `1`\n",
    "- Opposite vectors → similarity = `-1`\n",
    "- Perpendicular vectors → similarity = `0`\n",
    "- Scaled vectors (same direction, different length) → still `1`, because cosine similarity ignores magnitude and only cares about **direction**. This is why it's ideal for comparing text embeddings, which can have different lengths."
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "### 10. Key Takeaways (Recap)\n",
    "\n",
    "| Term | Meaning |\n",
    "|---|---|\n",
    "| Dot Product | Sum of element-wise multiplication of two vectors |\n",
    "| Magnitude | Length of a vector, from the origin |\n",
    "| Cosine Similarity | Dot product divided by product of magnitudes |\n",
    "| Range | -1 (opposite) to 1 (identical direction) |\n",
    "| Why it matters | Compares *direction/meaning*, ignores vector length — perfect for text embeddings |\n",
    "\n",
    "### 11. Practice Exercises\n",
    "1. Write `cosine_similarity_manual` using list comprehensions instead of loops.\n",
    "2. Calculate cosine similarity between `[1, 1, 1]` and `[5, 5, 5]` — predict the answer before running.\n",
    "3. Combine this with the Hugging Face embeddings notebook: replace `vector_A` and `vector_B` with real sentence embeddings and compare results."
   ]
  },
  {
   "cell_type": "markdown",
   "id": "ad062553",
   "metadata": {},
   "source": []
  }
 ],
 "metadata": {
  "colab": {
   "provenance": []
  },
  "kernelspec": {
   "display_name": "Python 3",
   "language": "python",
   "name": "python3"
  },
  "language_info": {
   "name": "python",
   "version": "3.10"
  }
 },
 "nbformat": 4,
 "nbformat_minor": 5
}
```
