```python
!pip install -q scikit-learn gensim scipy

from sklearn.preprocessing import OneHotEncoder
from gensim.models import Word2Vec

words = [['cat'], ['dog'], ['apple'], ['car']]

encoder = OneHotEncoder(sparse_output=False)
result = encoder.fit_transform(words)

print("Words:", encoder.categories_[0])
print("\nOne-Hot Embeddings:")
print(result)

sentences = [
    ["python", "is", "a", "language"],
    ["java", "is", "a", "language"],
    ["python", "is", "easy"],
    ["java", "is", "easy"],
    ["python", "is", "powerful"],
    ["java", "is", "powerful"]
]

model = Word2Vec(
    sentences=sentences,
    vector_size=10,
    window=2,
    min_count=1,
    workers=1,
    seed=42
)

print("\nWords similar to 'java':")
print(model.wv.most_similar("java", topn=3))

print("\nWords similar to 'python':")
print(model.wv.most_similar("python", topn=4))
```
